```python
from dataset import get_loader
import os
path = '../dataset'
if not os.path.exists(path):
    print("Dataset not found")
    exit()
data_loader = get_loader(path)
```

```python
import matplotlib.pyplot as plt
import torch
def draw_image(light_img, low_img):
    plt.figure(figsize= (24,12))
    plt.subplot(1,2,1)
    plt.imshow(light_img)
    plt.title("LIGHT IMAGE")
    plt.subplot(1,2,2)
    plt.imshow(low_img)
    plt.title("LOW IMAGE")
    plt.show()

low, light = next(iter(data_loader))
low = low.permute(0,2,3,1)
light = light.permute(0,2,3,1)
for low_img, light_img in zip(low, light):
    draw_image(light_img, low_img)
```
